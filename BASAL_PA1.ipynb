# A. WORD ROTATION PROBLEM

## Requirement: Use string indexing or slicing to construct the returned string.

```python
def rotate_word(txt):

    #Rotating a non-empty string by moving the first character to the end while keeping the rest in order.
    
    if not txt:
        return txt
    #Move the first character to the end
    return txt[1:] + txt[0]
    
#Main Program
#User input
user_txt = input("Enter a word: ")

#Call the function
result = rotate_word(user_txt)

#Output
print("Rotate word: ", result)
```

## B. USERNAME BUILDER PROBLEM

## Requirement: Use basic string methods and string concatenation. Return the completed username.

```python
def make_username(first_name, last_name):
    # Convert to lowercase and remove spaces from both names
    clean_first = first_name.lower().replace(" ", "")
    clean_last = last_name.lower().replace(" ", "")

    # Join processed names with a period
    return f"{clean_first}.{clean_last}"


# User Input Section
first_name_input = input("Enter first name: ")
last_name_input = input("Enter last name: ")

# Execute function and print returned result
username = make_username(first_name_input, last_name_input)
print("Generated Username:", username)
```

## C. BOOKEND SWAP PROBLEM

## Requirement: Use extended sequence unpacking in the following form: first, *middle, last = items

```python
def swap_bookends(items):
    # Unpack using extended sequence unpacking
    first, *middle, last = items

    # Return new list with first and last swapped
    return [last, *middle, first]


# User Input Section
user_input = input(
    "Enter list items separated by spaces (minimum 2 items): "
)
items_input = user_input.split()

# Execute function and print returned result
result = swap_bookends(items_input)
print("Resulting List:", result)
```
